```python
def recommed(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]

    # Calculate similarity only for this movie
    distances = cosine_similarity(
        vectors[movie_index],
        vectors
    ).flatten()

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```
